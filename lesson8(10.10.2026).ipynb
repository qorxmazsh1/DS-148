```python
letters = ['a', 'a', 'b', 'c', 'c', 'd', 'd', 'd', 'd']
count_l  = {}
def paltaryuyan(letters):
    for letter in letters:
        if letter not in count_l:
            count_l[letter] = letters.count(letter)
    return count_l

print(paltaryuyan(letters))
#{a: 2, b : 1 ....}
```

```python
number = int(input("Enter a number: "))
copy = number
reverse_number = 0
def is_palindrome(number):
    global copy, reverse_number
    while number > 0:
        digit = number % 10
        reverse_number = reverse_number * 10 + digit
        number //= 10
    return copy == reverse_number

is_palindrome(number)
```
