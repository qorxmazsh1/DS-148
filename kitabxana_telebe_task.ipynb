```python
## Tapşırıq 3: `Uzv` class-ı

- `limit` class atributu: eyni anda götürə biləcəyi kitab sayı (`3`)
- `ad` və `kitablar` (boş siyahı) atributları
- `say` property-si: üzvdəki kitab sayını qaytarır
- `limit_dolub()` metodu: limit dolubsa `True` qaytarır
```

```python
## Tapşırıq 4: `Kitabxana` class-ı

Atributlar: `ad`, `kitablar`, `uzvler` (hamısı başlanğıcda boş siyahı, `ad` istisna).

Metodlar:
- `kitab_elave_et(kitab)` və `uzv_qeydiyyat(uzv)`
- `axtar(soz)`: başlıq və ya müəllifdə söz keçən kitabların siyahısını qaytarır (böyük/kiçik hərf fərqi olmadan)
- `kitab_ver(uzv, basliq)`: kitabı üzvə verir. Kitab tapılmasa, artıq götürülübsə və ya üzvün limiti dolubsa `ValueError` qaldır
- `qaytar(uzv, basliq)`: kitabı geri qaytarır. Kitab həmin üzvdə deyilsə `ValueError` qaldır
```

```python
## Tapşırıq 5: Varislik

`Uzv`-dən miras alan iki class yaz:
- `TelebeUzv`: limit `5`
- `MuellimUzv`: limit `10`

Başqa heç nə yazmağa ehtiyac yoxdur, yalnız class atributunu dəyiş.
```
