```python
## Tapşırıq 4: `Kitabxana` class-ı

Atributlar: `ad`, `kitablar`, `uzvler` (hamısı başlanğıcda boş siyahı, `ad` istisna).

Metodlar:
- `kitab_elave_et(kitab)` və `uzv_qeydiyyat(uzv)`
- `axtar(soz)`: başlıq və ya müəllifdə söz keçən kitabların siyahısını qaytarır (böyük/kiçik hərf fərqi olmadan)
- `kitab_ver(uzv, basliq)`: kitabı üzvə verir. Kitab tapılmasa, artıq götürülübsə və ya üzvün limiti dolubsa `ValueError` qaldır
- `qaytar(uzv, basliq)`: kitabı geri qaytarır. Kitab həmin üzvdə deyilsə `ValueError` qaldır
```

```python
## Tapşırıq 5: Varislik

`Uzv`-dən miras alan iki class yaz:
- `TelebeUzv`: limit `5`
- `MuellimUzv`: limit `10`

Başqa heç nə yazmağa ehtiyac yoxdur, yalnız class atributunu dəyiş.
```

```python
## Bonus

`Kitabxana` class-ına əlavə et:
- `__len__`: kitabxanadakı ümumi kitab sayı
- `movcud_kitablar` property-si: yalnız götürülməmiş kitabların siyahısı
```

```python
class Kitabxana:
    def __init__(self, uzv):
        self.uzv = uzv
        self.kitablar = [] #--> dict-e deyis ve saygac yaz kitablari saysin
        self.uzvler = [] 
    def uzv_qeydiyyat(self, uzv): #--> args istifade et
        self.uzvler.append(uzv)
    def kitab_elave_et(self, kitab): #-->args istifade et 
        self.kitablar.append(kitab)

kitabxana = Kitabxana('Kitabxana')
kitabxana.uzv_qeydiyyat('Eltac')
kitabxana.kitab_elave_et('heyvan fermasi')
print(kitabxana.uzvler)  # ['Eltac']
print(kitabxana.kitablar)  # ['heyvan fermasi']
```
