```python
nums = [1, 2, 2, 3, 3, 3, 4, 5, 5]


#output >> [1, 2, 3, 4, 5]
def unique_numbers(nums):
    uni_list = []
    for num in nums:
        if num not in uni_list:
            uni_list.append(num)
    return uni_list

print(nums)
```
